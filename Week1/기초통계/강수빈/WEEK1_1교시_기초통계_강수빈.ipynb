## Q2-2. 실험에 필요한 최소 test users 수를 구해봅시다.
유의수준 0.05, 검정력 80% 하에 수행해주세요.  


아래 공식 참고해서, 최소 test users 수를 구하는 코드 작성해주세요.

추가적으로 n은 산출하는데에 가정한 p1, p2도 같이 작성해주세요.

$$n = \frac{(Z_{\alpha/2} + Z_{\beta})^2 \cdot [p_1(1-p_1) + p_2(1-p_2)]}{(p_1 - p_2)^2}$$

## A2-2. 아래 질문에 대한 답과 코드를 채워주세요.

p1, p2는 각각 무엇을 의미하는 확률이고, 어떤 값으로 가정하셨나요?   
또한 위 공식을 보고 n을 구하기 위한 코드를 작성해주세요. 단, 라이브러리는 z값을 구하는 용도의 것만 사용해주세요.


### 답변:

p1: 대조군(A, 기존 UI)에서의 주지표 성공확률

*   여기서는 추천 카드 클릭률(Recommendation CTR) 을 주지표로 두고,
*   p1 = 0.12 (12%) 로 가정 (추천 근거가 없어서 클릭률이 낮은 상태)



p2: 실험군(B, 추천 근거 표시 UI)에서의 주지표 성공확률

*   추천 근거를 제공하면 신뢰/이해가 올라 클릭이 늘어난다고 보고,
*   p2 = 0.15 (15%) 로 가정 (절대 +3%p 개선 가정)





즉, p1/p2는 “각 그룹에서 사용자가 추천 카드를 클릭할 확률(CTR)”입니다.

```python
# A2-2 코드작성. 나온 n은 아래 Minimum Sample Size에 기재.
import math
from scipy.stats import norm

# 1) 실험 설정
alpha = 0.05          # 유의수준
power = 0.80          # 검정력 = 1 - beta
beta = 1 - power

# 2) p1, p2 가정 (주지표: 추천 카드 클릭률 CTR)
p1 = 0.12  # control CTR
p2 = 0.15  # variant CTR

# 3) Z값 계산
z_alpha_over_2 = norm.ppf(1 - alpha/2)  # two-sided
z_beta = norm.ppf(1 - beta)

# 4) 샘플 사이즈 공식
n = ((z_alpha_over_2 + z_beta)**2 * (p1*(1-p1) + p2*(1-p2))) / ((p1 - p2)**2)

# 5) 올림 처리 (그룹당 최소 인원)
n_per_group = math.ceil(n)
total_n = n_per_group * 2

print("z_alpha/2 =", z_alpha_over_2)
print("z_beta    =", z_beta)
print("p1 =", p1, ", p2 =", p2)
print("n (per group) =", n_per_group)
print("Total sample size =", total_n)
```

[실험 가설]  
귀무가설 ($H_0$): 메인 홈 검색 모듈의 위치(하단 vs 최상단)와 상세 페이지 진입 여부 사이에는 상관관계가 없다. 즉, 두 그룹의 상세 페이지 진입률($CVR$)은 동일하다.$$H_0: p_A = p_B$$대립가설 ($H_1$): 메인 홈 검색 모듈을 최상단에 노출하는 것은 상세 페이지 진입 여부에 영향을 미친다. 즉, 실험군(B)의 상세 페이지 진입률이 대조군(A)보다 높을 것이다.$$H_1: p_A < p_B$$

## A3-1.  검정 이름과 이유를 설명해주세요.

답변: 위 가설에 필요한 검정은 카이제곱 검정 이고,  

이유는 상세 페이지 진입 여부가 ‘성공/실패’ 형태의 범주형 데이터이며, 검색바 위치(A/B)에 따른 두 집단 간 전환율(CVR) 차이가 통계적으로 유의미한지를 2×2 분할표를 통해 검정할 수 있기 때문이다. 또한 본 실험은 비율 비교를 목적으로 하므로 연속형 평균 비교에 사용되는 t-test보다 카이제곱 검정이 더 적합하다.

<br>

```python
'''''''''''''''''''''''''''''''''''''''
제가 세운 A/B TEST 설계 실험 코드입니다.
'''''''''''''''''''''''''''''''''''''''

import pandas as pd
from scipy.stats import chi2_contingency

# 1. 관측 데이터 입력 (실제 실험 결과 수치로 변경하세요)
# [상세진입 성공 수, 상세진입 실패 수]
control_results = [11500, 34500]  # 대조군(A)
variant_results = [12190, 33810]  # 실험군(B)

# 2. 분할표(Contingency Table) 생성
data = [control_results, variant_results]

# 3. 카이제곱 검정 수행
chi2, p_value, dof, expected = chi2_contingency(data)

# 4. 결과 출력
print(f"카이제곱 통계량: {chi2:.4f}")
print(f"P-value: {p_value:.10f}")

alpha = 0.05
if p_value < alpha:
    print(f"결과: P-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.")
    print("=> 검색바 최상단 노출이 상세 페이지 진입률에 유의미한 변화를 주었습니다.")
else:
    print(f"결과: P-value가 {alpha}보다 크므로 귀무가설을 기각할 수 없습니다.")
    print("=> 검색바 위치 변경의 효과가 통계적으로 증명되지 않았습니다.")
```

귀무가설 ($H_0$): 추천 결과에 대한 근거 표시 여부(A: 미표시, B: 표시)는 추천 카드 클릭 여부와 상관관계가 없다.
즉, 두 그룹의 추천 클릭률(CTR)은 동일하다.$$H_0: p_A = p_B$$대립가설 ($H_1$): 추천 결과에 대한 근거를 표시하면 추천 카드 클릭률이 증가한다.
즉, 실험군(B)의 CTR이 대조군(A)보다 높다.$$H_1: p_A < p_B$$

이유는 클릭 여부 및 전환 여부가 성공/실패 형태의 범주형 데이터이며, A/B 두 집단 간 전환 비율의 차이가 통계적으로 유의미한지를 2×2 분할표를 통해 검정할 수 있기 때문이다. t-test는 연속형 데이터의 평균 비교에 사용되므로 본 실험에는 적합하지 않다.

## Q3-4. 위에서 답한 것을 토대로, 검정을 수행해주세요. (t test와 카이제곱 검정 중)


## A3-4 아래 셀에 코드 작성해주세요. 그리고!!!
그리고 나온 값을 통해 수행하신 a/b test 결과를 설정하신 가설에 맞게 해석해주세요.

```python
import pandas as pd
from scipy.stats import chi2_contingency

# 1. 관측 데이터 입력 (추천 근거 실험 결과)
# [추천 클릭 수, 추천 미클릭 수]

# A: 추천 근거 미표시 (Control)
control_results = [820, 6180]

# B: 추천 근거 표시 (Variant)
variant_results = [980, 6020]

# 2. 분할표(Contingency Table) 생성
data = [control_results, variant_results]

# 3. 카이제곱 검정 수행
chi2, p_value, dof, expected = chi2_contingency(data)

# 4. 결과 출력
print("=== 추천 근거 표시 A/B Test 결과 ===")
print(f"Contingency Table: {data}")
print(f"카이제곱 통계량: {chi2:.4f}")
print(f"P-value: {p_value:.10f}")

alpha = 0.05

if p_value < alpha:
    print(f"결과: P-value가 {alpha}보다 작으므로 귀무가설을 기각합니다.")
    print("=> 추천 근거 표시가 추천 클릭률(CTR)을 유의미하게 향상시켰습니다.")
else:
    print(f"결과: P-value가 {alpha}보다 크므로 귀무가설을 기각할 수 없습니다.")
    print("=> 추천 근거 표시의 효과가 통계적으로 증명되지 않았습니다.")
```

**결과해석**

답변:
카이제곱 검정 결과, p-value는 0.00005954로 유의수준 0.05보다 매우 작게 나타났다.
따라서 귀무가설(H₀: 추천 근거 표시 여부에 따른 클릭률 차이가 없다)을 기각한다.
이는 추천 결과에 대한 근거를 함께 제공한 실험군(B)이 대조군(A)보다 통계적으로 유의미하게 높은 클릭률(CTR)을 보였음을 의미한다.
즉, 추천 근거 표시 기능은 사용자의 추천 신뢰도를 높이고, 추천 콘텐츠에 대한 클릭 행동을 유의미하게 증가시키는 효과가 있는 것으로 판단된다.
